```python
import os
import json
from tqdm import tqdm
```

```python
foldername="/proj/m3benchmark/m3data/0828/balanced_rest_v2_generate/final"
output_foldername="/proj/m3benchmark/m3data/0828/balanced_rest_v2_generate_with_domain/final"
```

```python
filenames=os.listdir(foldername)
for filename in tqdm(filenames):
    domain=filename.split("_multiturn_bird")[0]
    with open(f"{foldername}/{filename}", 'r') as j:
        data = json.load(j)
    for idx, item in enumerate(data):
        data[idx]["domain"]=domain
    with open(f'{output_foldername}/{filename}', 'w') as fout:
        json.dump(data , fout)
```
