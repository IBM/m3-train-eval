```python
foldernames=[
    ("/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_chunked/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_after_generate/final/","/proj/m3benchmark/m3data/0905/balanced_rest_v4_exploratory_trajectory/trajectories/","/proj/m3benchmark/m3data/0905/expert_vs_agent_dpo_pairs/train_multi_turn_no_scenarios.json"), # [TRAIN] Multi-turn no scenarios
    ("/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_chunked_scenarios/","/proj/m3benchmark/danish/m3data/0905/m3_train_test_ood_rest_v2_chunked_scenarios_gt/final/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_chunked_scenarios_exp/trajectories/","/proj/m3benchmark/m3data/0905/expert_vs_agent_dpo_pairs/train_multi_turn_with_scenarios.json"), # [TRAIN] Multi-turn scenarios
    ("/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_single_turn_chunked/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_single_turn_after_generate/final/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_single_turn_exploratory_trajectory/trajectories/","/proj/m3benchmark/m3data/0905/expert_vs_agent_dpo_pairs/train_single_turn_no_scenarios.json"), # [TRAIN] Single Turn no scenarios
    ("/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_single_turn_chunked_scenarios/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_single_turn_chunked_scenarios_st_gt/final/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2_single_turn_chunked_scenarios_st_exp/trajectories/","/proj/m3benchmark/m3data/0905/expert_vs_agent_dpo_pairs/train_single_turn_with_scenarios.json"), # [TRAIN] Single Turn scenarios
    ("/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2/test_chunked/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2/test_chunked_gt/final/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2/test_chunked_gt_exp/trajectories/","/proj/m3benchmark/m3data/0905/expert_vs_agent_dpo_pairs/test_multi_turn_no_scenarios.json"),  # [TEST] Multi-turn no scenarios
    ("/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2/test_scenarios_chunked/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2/test_scenarios_chunked_gt/final/","/proj/m3benchmark/m3data/0905/m3_train_test_ood_rest_v2/test_scenarios_chunked_exp/trajectories","/proj/m3benchmark/m3data/0905/expert_vs_agent_dpo_pairs/test_multi_turn_with_scenarios.json"), # [TEST] Multi-turn scenarios
]
```

```python
row_names = ['[TRAIN][multi-turn][no-scenarios]', '[TRAIN][multi-turn][scenarios]', '[TRAIN][single-turn][no-scenarios]','[TRAIN][single-turn][scenarios]','[TEST][multi-turn][no-scenarios]', '[TEST][multi-turn][scenarios]']
column_names = ["samples","ground_truth","exploratory_trajectory","dpo_trajectory"]
stats = pd.DataFrame(columns=column_names, index=row_names)
for idx, (base, gt, et, dpo) in enumerate(foldernames):
    num_base_examples, num_gt, num_et, num_dpo=0, 0, 0, 0

    # Base dataset count
    filenames=os.listdir(base)
    for filename in filenames:
        with open(f"{base}/{filename}", 'r') as j:
            data = json.load(j)
        num_base_examples+=len(data)
    
    # Ground Truth Trajectories count
    filenames=os.listdir(gt)
    for filename in filenames:
        if "_final.json" in filename:
            with open(f"{gt}/{filename}", 'r') as j:
                data = json.load(j)
            num_gt+=len(data)

    # Exploratory Trajectories count
    filenames=os.listdir(et)
    num_et=len(filenames)

    # DPO Trajectories count
    with open(f"{dpo}", 'r') as j:
        data = json.load(j)
    num_dpo=len(data)

    stats.loc[row_names[idx]] = [num_base_examples, num_gt, num_et, num_dpo]
    print([num_base_examples, num_gt, num_et, num_dpo])
```
